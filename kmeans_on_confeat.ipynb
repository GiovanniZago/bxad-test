```python
from sklearn.cluster import KMeans
import numpy as np
import uproot
import awkward as ak
import matplotlib.pyplot as plt
import pickle
```

```python
filepath = "data/output_1000_seq3s.root"
data = uproot.open(filepath + ":L1BMTFStubSequences").arrays()
df = ak.to_dataframe(data)
```

```python
hweta = df["L1BMTFStub_hwEta"].to_numpy().reshape((-1, 1))

hweta_kmeans = KMeans(n_clusters=9, random_state=0).fit(hweta)
counts, bins, _ = plt.hist(df["L1BMTFStub_hwEta"].to_numpy());
plt.vlines(hweta_kmeans.cluster_centers_, ymin=0, ymax=max(counts), color="r");
print(np.sort(hweta_kmeans.cluster_centers_))

with open("hweta_kmeans.pkl", "wb") as fp:
    pickle.dump(hweta_kmeans, fp)
```

```python
hwqeta = df["L1BMTFStub_hwQEta"].to_numpy().reshape((-1, 1))

hwqeta_kmeans = KMeans(n_clusters=6, random_state=0).fit(hwqeta)
counts, bins, _ = plt.hist(df["L1BMTFStub_hwQEta"].to_numpy());
plt.vlines(hwqeta_kmeans.cluster_centers_, ymin=0, ymax=max(counts), color="r");
print(np.sort(hwqeta_kmeans.cluster_centers_))

with open("hwqeta_kmeans.pkl", "wb") as fp:
    pickle.dump(hwqeta_kmeans, fp)
```
